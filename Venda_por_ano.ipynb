```python
import pandas as pd
from IPython.display import display as ds

```

```python
caminho_vendas2020 = r'Aula 1 - Importando Base de Dados\Base Vendas - 2020.xlsx'
caminho1 = r'Aula 1 - Importando Base de Dados\Base Vendas - 2021.xlsx'
caminho2 = r'Aula 1 - Importando Base de Dados\Base Vendas - 2022.xlsx'
caminho3 = r'Aula 1 - Importando Base de Dados\Cadastro Lojas.xlsx'
caminho_devol = r'Aula 1 - Importando Base de Dados\Base Devoluções.xlsx'

df_2020 = pd.read_excel(caminho_vendas2020)
df_2021 = pd.read_excel(caminho1)
df_2022 = pd.read_excel(caminho2)
```

```python
ds(df_2020)
```

```python
print(df_2020.isnull().sum())
print(df_2020.info())
print(f'\n{'=-'*30}\n')
print(df_2021.isnull().sum())
print(df_2021.info())
print(f'\n{'=-'*30}\n')
print(df_2022.isnull().sum())
print(df_2022.info())
print(f'\n{'=-'*30}\n')
```

Os IDs vieram como int, o que pode tirar o 0 da frente em algum caso, mas como todos vieram assim
o id vai ficar compatível


```python
df_2020_mergedby_loja = df_2020.groupby('ID Loja')['Qtd Vendida'].sum().reset_index()
ds(df_2020_mergedby_loja)
```

```python
df_2021_mergedby_loja = df_2021.groupby('ID Loja')['Qtd Vendida'].sum().reset_index()
ds(df_2021_mergedby_loja)
```

```python
df_2022_mergedby_loja = df_2022.groupby('ID Loja')['Qtd Vendida'].sum().reset_index()
ds(df_2022_mergedby_loja)
```

```python
todos = df_2020_mergedby_loja.merge(df_2021_mergedby_loja,  on='ID Loja').merge(df_2022_mergedby_loja, on='ID Loja')
todos = todos.rename(columns={'Qtd Vendida_x': 'Vendas_2020',
    'Qtd Vendida_y': 'Vendas_2021',
    'Qtd Vendida':   'Vendas_2022'})

ds(todos)
```

```python
df_lojas = pd.read_excel(caminho3)
ds(df_lojas)
```

```python
df_lojas = df_lojas[['ID Loja','Nome da Loja','Quantidade Colaboradores', 'Tipo', 'Gerente Loja']]
df_lojas['Gerente Loja'] = df_lojas['Gerente Loja'].str.split(', ').str[::-1].str.join(' ')
print(df_lojas)
```

```python
df_2020final = df_2020_mergedby_loja.merge(df_lojas, on='ID Loja', how='left').sort_values(by='Qtd Vendida', ascending=False)
df_2020final['Valor - Funcionarios'] = (df_2020final['Qtd Vendida']/ df_2020final['Quantidade Colaboradores'])
df_2021final = df_2021_mergedby_loja.merge(df_lojas, on='ID Loja', how='left').sort_values(by='Qtd Vendida', ascending=False)
df_2021final['Valor - Funcionarios'] = (df_2021final['Qtd Vendida']/ df_2021final['Quantidade Colaboradores'])
df_2022final = df_2022_mergedby_loja.merge(df_lojas, on='ID Loja', how='left').sort_values(by='Qtd Vendida', ascending=False)
df_2022final['Valor - Funcionarios'] = (df_2022final['Qtd Vendida']/ df_2022final['Quantidade Colaboradores'])
ds(df_2020final)
ds(df_2021final)
ds(df_2022final)
```

```python
df_2020final['Ano'] = 2020
df_2021final['Ano'] = 2021
df_2022final['Ano'] = 2022
with pd.ExcelWriter('Vendas por ano.xlsx', engine='openpyxl') as writer:
    df_2020final.to_excel(writer, sheet_name='2020', index=False)
    df_2021final.to_excel(writer, sheet_name='2021', index=False)
    df_2022final.to_excel(writer, sheet_name='2022', index=False)
```

```python
df_devol = pd.read_excel(caminho_devol)
df_devol['Data Devolução'] = df_devol['Data Devolução'].dt.year
ds(df_devol.info())
ds(df_devol)
```

```python
df_devol2020 = df_devol[df_devol['Data Devolução'] == 2020]
df_devol2021 = df_devol[df_devol['Data Devolução'] == 2021]
df_devol2022 = df_devol[df_devol['Data Devolução'] == 2022]

df_devol2020 = df_devol2020.sort_values(by='Qtd Devolvida', ascending=False)
df_devol2021 = df_devol2021.sort_values(by='Qtd Devolvida', ascending=False)
df_devol2022 = df_devol2022.sort_values(by='Qtd Devolvida', ascending=False)

df_devol2020 = df_devol2020.merge(df_lojas, on='ID Loja', how='left')
df_devol2021 = df_devol2021.merge(df_lojas, on='ID Loja', how='left')
df_devol2022 = df_devol2022.merge(df_lojas, on='ID Loja', how='left')
ds(df_devol2020)
ds(df_devol2021)
ds(df_devol2022)
```

```python
with pd.ExcelWriter('Devoluções por ano.xlsx', engine='openpyxl') as writer:
    df_devol2020.to_excel(writer, sheet_name='2020', index=False)
    df_devol2021.to_excel(writer, sheet_name='2021', index=False)
    df_devol2022.to_excel(writer, sheet_name='2022', index=False)
```

```python
por_data2020 = df_2020.groupby(['Data da Venda', 'ID Loja']).agg(Total_Vendido=('Qtd Vendida', 'sum')).reset_index()
por_data2020['Data da Venda'] = por_data2020['Data da Venda'].dt.normalize()
por_data2020['Dia'] = por_data2020['Data da Venda'].dt.day
por_data2020['Mes'] = por_data2020['Data da Venda'].dt.month
por_data2020['Ano'] = por_data2020['Data da Venda'].dt.year
por_data2020.drop(columns=['Data da Venda'], inplace=True)
ds(por_data2020)
```

```python
por_data2020 = por_data2020.merge(df_lojas, on='ID Loja', how='left')
ds(por_data2020)
```

```python
por_data2021 = df_2021.groupby(['Data da Venda', 'ID Loja']).agg(Total_Vendido=('Qtd Vendida', 'sum')).reset_index()
por_data2021['Data da Venda'] = por_data2021['Data da Venda'].dt.normalize()
por_data2021['Dia'] = por_data2021['Data da Venda'].dt.day
por_data2021['Mes'] = por_data2021['Data da Venda'].dt.month
por_data2021['Ano'] = por_data2021['Data da Venda'].dt.year
por_data2021.drop(columns=['Data da Venda'], inplace=True)
por_data2021 = por_data2021.merge(df_lojas, on='ID Loja', how='left')
ds(por_data2021)
```

```python
por_data2022 = df_2022.groupby(['Data da Venda', 'ID Loja']).agg(Total_Vendido=('Qtd Vendida', 'sum')).reset_index()
por_data2022['Data da Venda'] = por_data2022['Data da Venda'].dt.normalize()
por_data2022['Dia'] = por_data2022['Data da Venda'].dt.day
por_data2022['Mes'] = por_data2022['Data da Venda'].dt.month
por_data2022['Ano'] = por_data2022['Data da Venda'].dt.year
por_data2022.drop(columns=['Data da Venda'], inplace=True)
por_data2022 = por_data2022.merge(df_lojas, on='ID Loja', how='left')
ds(por_data2022)
```

```python
with pd.ExcelWriter('Total Vendas.xlsx', engine='openpyxl') as writer:
    por_data2020.to_excel(writer, sheet_name='2020', index=False)
    worksheet = writer.sheets['2020']
    worksheet.auto_filter.ref = worksheet.dimensions
    por_data2021.to_excel(writer, sheet_name='2021', index=False)
    worksheet = writer.sheets['2021']
    worksheet.auto_filter.ref = worksheet.dimensions
    por_data2022.to_excel(writer, sheet_name='2022', index=False)
    worksheet = writer.sheets['2022']
    worksheet.auto_filter.ref = worksheet.dimensions
```
